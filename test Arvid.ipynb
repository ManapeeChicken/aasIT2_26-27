Oppg9)
a)
x*(sirkel["s1"]/2) regner ut omkretsen til sirkel 1, og ved å lage slike funksjoner for sirkel 2 og sirkel 3 kan vi finne ut av hvilke sirkler som har lengst omkrets. (er ganske ubrukelig, siden selvfølgelig vil sirkelen med størst areal ha lengst omkrets)

```python
#Oppg9) 
#b)
sirkel = {"s1":float(input("radiusen til sirkel 1 i cm: ")), #finner ut av rediusen til sirkel 1
          "s2":float(input("radiusen til sirkel 2 i cm: ")), #etc
          "s3":float(input("radiusen til sirkel 3 i cm: ")) #etc
          }

x = 3.14

sirkel1 = x* sirkel["s1"]**2 #regner ut arealet til sirkel 1
sirkel2 = x* sirkel["s2"]**2 #etc
sirkel3 = x* sirkel["s3"]**2 #etc

omkrets1 = x*(sirkel["s1"]/2) #regner ut omkretsen til sirkel 1
omkrets2 = x*(sirkel["s2"]/2) #etc
omkrets3 = x*(sirkel["s3"]/2) #etc

if sirkel1 > sirkel2 and sirkel1 > sirkel3: #finner ut hvilken sirkel som er størst
    print('{sirkel1} er størst')
elif sirkel2 > sirkel3:
    print('{sirkel2} er størst')
elif sirkel3 > sirkel2:
    print('{sirkel3} er størst')

if omkrets1 > omkrets2 and omkrets1 > omkrets3: #finner ut hvilken omkrets som er størst
    print('{omkrets1} er størst')
elif omkrets2 > omkrets3:
    print('{omkrets2} er størst')
elif omkrets3 > omkrets2:
    print('{omkrets3} er størst')
```
