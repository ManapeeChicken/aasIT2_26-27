```python
list = [100, 120, 125, 155, 160]

for i in list:
    if i<115:
        print(i,"cm", "du er ikke høy nok for å kjøre")
    elif i<125:
        print(i,"cm", "du er ikke høy nok for å kjøre selv")
    elif i<=155:
        print(i,"cm", "du er høy nok for å kjøre alene")
```
